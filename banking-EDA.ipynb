**Income Band Analysis:**
The majority of customers belong to the **Medium Income Band**, followed by the **Low Income Band**, while the **High Income Band represents the smallest customer segment**. This indicates that the bank's customer portfolio is primarily concentrated in middle-income individuals, providing significant opportunities for targeted retail banking products and services.

```python
bins = [0, 100000, 300000, float('inf')]
labels = ['Low', 'Med', 'High']

# Apply pd.cut to categorize income levels
df['Income Band'] = pd.cut(
    df['Estimated Income'],
    bins=bins,
    labels=labels,
    right=False
)
```

```python
df['Income Band'].value_counts().plot(kind='bar')
```

Numerical Anlysis

```python
numerical_cols=['Estimated Income','Superannuation Savings','Credit Card Balance','Bank Loans','Bank Deposits','Checking Accounts','Saving Accounts','Foreign Currency Account', 'Business Lending']
#univariate analysis
plt.figure(figsize=(15, 15)) # Adjust figure size for 9 plots
for i,col in enumerate(numerical_cols):
  plt.subplot(3, 3, i+1) # Changed to 3 rows, 3 columns
  sns.histplot(df[col],kde=True)
  plt.title(f'Distribution of {col}')
plt.tight_layout() # Adjust layout to prevent overlapping titles
plt.show()
```

• Most financial variables exhibit a **right-skewed distribution**, with a large concentration of customers in lower-to-mid value ranges.

**• Estimated Income** is concentrated among middle-income customers, while a smaller number of customers earn significantly higher incomes.

**• Bank Deposits and Bank Loans** show similar patterns, where most customers maintain moderate balances and only a few hold very large amounts.

**• Checking and Saving Accounts** are heavily concentrated at lower balances, indicating that the majority of customers maintain standard account balances.

•**Credit Card Balance is positively skewed** , suggesting that only a limited number of customers carry high outstanding balances.

**• Business Lending and Foreign Currency Accounts** show a long-tail distribution, with a small group of customers contributing a substantial share of total value.

• The presence of extreme values across several banking products indicates the existence of **high-value customers** within the portfolio.

**Correlation Analysis – Key Insights**

Heatmaps


```python
numerical_cols=['Estimated Income','Superannuation Savings','Credit Card Balance','Bank Loans','Bank Deposits','Checking Accounts','Saving Accounts','Foreign Currency Account', 'Business Lending']
correlation_matrix = df[numerical_cols].corr()
plt.figure(figsize=(10, 8))
sns.heatmap(correlation_matrix, annot=True, cmap='crest', linewidths=0.5)
plt.title('Correlation Heatmap')
```

**Key Insights**

• Bank Deposits and Checking Accounts show the strongest **positive correlation (0.84)**, indicating customers with higher deposits tend to maintain larger checking balances.

**•Bank Deposits and Saving Accounts** are highly correlated (0.75), highlighting strong customer engagement across deposit products.

• Business Lending has moderate positive relationships with **Bank Deposits (0.44) and Bank Loans (0.42)**, suggesting high-value customers utilize multiple banking services.

• Credit Card Balance is positively associated with both **Bank Deposits (0.38) and Bank Loans (0.37)**, reflecting increased credit usage among financially active customers.

• Foreign Currency Accounts exhibit moderate correlations with deposits, loans, and business lending, indicating higher engagement from international banking customers.

• Estimated Income shows only **moderate correlations (0.26–0.33)** with banking products, suggesting customer behavior depends on factors beyond income alone.

**• No negative correlations** are observed among key variables, indicating that growth in one banking product is generally associated with increased usage of other banking services.
